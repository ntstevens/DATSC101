## What is Data Science?

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/what_is_ds_joke.jpeg" alt="ds dino" width="800"/>

[Image Source](https://programmerhumor.io/math-memes/the-evolutionary-tale-of-a-data-scientist-0d2f)

_Data science_ is an interdisciplinary blend of mathematics, statistics and computer science that applies scientific methods to extract information and provide insight from data. Interest lies in the use and development of tools and theories for the principled collection and systematic study of data with emphasis on problem solving, knowledge generation and advancement in all domains for which data drives decision-making.

Let's break this down a little bit...

* Data scientists use a combination of math, statistics, and computer science
* Data scientists carefully and systematically collect data
* Data scientists carefully and systematically extract information from data
* Data scientists also create the tools and methods to do all of the above

**Main takeaway:** Data scientists are problem solvers who through data collection and analysis facilitate decision-making.

### ⚾️ Data Science in Sports: Moneyball

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/moneyball.jpg" alt="moneyball" width="500"/>
</div>

In the early 2000s, the Oakland Athletics baseball team faced a problem familiar to many small organizations: the have a fraction of the budget that rivals like the New York Yankees do, yet they have the same goal: win games.

General manager Billy Beane and his analysts sought to determine which player statistics _actually_ predicted wins.

By analyzing decades of performance data, they discovered that traditional scouting metrics like _batting average_ were poor predictors of scoring runs, while overlooked statistics like _on-base percentage_ were far more valuable and, crucially, undervalued in the player market.

This data-driven approach let the Oakland A's assemble a competitive roster for a fraction of the cost, sparking a revolution now used across nearly every professional sport.

### 🦠 COVID-19 in Canada: Epidemiological Modeling and Disease Surveillance

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/covid_cases_canada.png" alt="covid" width="800"/>
</div>

[Image Source](https://newsinteractives.cbc.ca/coronavirustracker/)

When COVID-19 arrived in Canada in early 2020, public health officials faced an urgent question with almost no precedent to draw on: how fast would the virus spread here, and which interventions would actually slow it down? 

In April 2020, Genome Canada launched CanCOGeN, a coast-to-coast network linking federal and provincial public health labs, hospitals, and universities, tasked with sequencing the genomes of up to 150,000 viral samples to track how the virus was spreading and mutating across the country. 

When the first troubling variants emerged in December 2020, a dedicated team of Canadian mathematical modellers began working directly with this genomic data to project how variants like Alpha, Delta, and later Omicron would likely spread, feeding directly into federal and provincial public health decisions. 

Later in the pandemic, as testing capacity in many provinces fell behind the sheer volume of Omicron-era infections, the Public Health Agency of Canada's National Microbiology Laboratory turned to an entirely different data source: wastewater. By measuring viral fragments shed by both symptomatic and asymptomatic people in sewage samples from communities across the country, researchers could estimate real community infection levels even when clinical testing alone could no longer keep up.

### 🌦️ Modeling Climate Change

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/climate%20modeling.png" alt="climate change" width="800"/>
</div>

[Image Source](https://science.nasa.gov/earth/climate-change/study-confirms-climate-models-are-getting-future-warming-projections-right/)

Given data on greenhouse gas emissions, ocean temperatures, ice coverage, and atmospheric patterns, how are global and regional climates likely to evolve over time? Which regions will face the greatest risk from sea level rise, drought, and extreme weather?

Understanding how Earth's climate will change over the coming decades is one of the most consequential data science challenges of our time. It's also one of the hardest challenges, since it requires making reliable predictions decades into the future about an enormously complex system. 

Researchers build and run climate models that combine mathematical equations with historical data to simulate the interactions between oceans, atmosphere, and land. 

Encouragingly, such models have proven strikingly accurate in predicting the warming trends we're now experiencing, validating this data-driven approach and increasing confidence in projections about the decades ahead.

### 🔥 Tracking and Predicting Wildfires in Canada

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/fire_map.png" alt="fires" width="800"/>
</div>

[Image Source](https://cwfis.cfs.nrcan.gc.ca/en/interactive-map)

Wildfires have become one of the defining environmental challenges facing Canada in particular, as has become apparent in recent summers. In 2023, roughly 15 million hectares burned, forcing over 230,000 people to evacuate. The smoke from these fires was experienced as far away as New York City. Something similar happened again in 2026. 

This raises two distinct data science questions: where are fires burning *right now*, and where is fire danger likely to *emerge* in the days and months ahead? 

The first question is answered largely from space. For instance, NASA uses satellites with heat-sensing instruments to detect active fire "hotspots" anywhere on Earth, often within hours of ignition. This is especially valuable in remote parts of Canada where there are vast but sparsely populated forest regions where there is no one to report a new fire. 

The second question is answered largely on the ground. Canada's Wildland Fire Information System combines data from roughly 2,500 weather stations across Canada and the US, informing predictive models that  forecast elevated fire risk months in advance. 

Together, these two data-driven systems form the backbone of how Canada monitors and prepares for a threat that is intensifying as the climate warms.

- [Canadian Wildland Fire Information System — Interactive Map](https://cwfis.cfs.nrcan.gc.ca/en/interactive-map) — Natural Resources Canada

### 🪐 Predicting Planets: Machine Learning and the Search for Exoplanets

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/light_curve.png" alt="light curve" width="800"/>
</div>

[Image Source](https://research.google/blog/open-sourcing-the-hunt-for-exoplanets/)

NASA's Kepler space telescope was (it's retired now) a planet-hunting observatory. Unlike telescopes that take deep-space pictures, Kepler used a specialized photometer to measure changes in *stellar brightness*.

Between 2009 and 2018, Kepler stared continuously at over 200,000 stars, measuring tiny dips in their brightness caused by orbiting planets passing in front of them — a technique called the transit method. 

In 2017, researchers at Google Brain and an astronomer from the University of Texas at Austin, developed a machine learning model to reliably identify light curves that suggested the existence of a planet. 

Over its operational life, the Kepler space telescope together with this data-driven approach, identified over 2,700 exoplanets, expanding human understanding of the cosmos.

### 🗺️ Mapping the Past: GIS and Historical Research

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/DECIMA.jpg" alt="DECIMA" width="800"/>
</div>

[Image Source](https://decima-map.net/how-to-use-decima-introduction/using-the-decima-webgis/)

Historians are usually associated with archives and old documents, not spatial data. But a growing field called historical GIS (Geographic Information Systems) encodes centuries-old records as mappable data, and the DECIMA project is one of the best examples. 

Led by researchers at the University of Toronto and Brock University, DECIMA takes three censuses of Florence, Italy from 1551, 1561, and 1632 and digitizes tens of thousands of individual household and occupation records. 

The team then georeferences this data onto a hand-drawn 1584 map of the city, plotting where individual butchers, silk weavers, notaries, and other merchants actually lived and worked down to the specific building. 

This turns static, hard-to-evaluate paper records into an interactive, queryable spatial dataset, letting historians ask (and answer) questions that plain reading of the archives could not handle efficiently. For instance:

* How did wealth and occupation cluster across neighborhoods?
* How did the city's population and economy shift after the 1630 plague? 
* What might a resident's daily sensory experience of walking through the city have actually been like? 

It's a clear demonstration that data science is not confined to STEM fields; the same core skills of structuring, linking, and analyzing data can transform how humanities scholars do their work.

- [DECIMA WebGIS (interactive map)](https://decima-map.net/how-to-use-decima-introduction/using-the-decima-webgis/)  — Official Site

### 🌉 Keeping Bridges Standing: Data Science in Structural Engineering

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/confederation_bridge.jpg" alt="confederation bridge" width="800"/>
</div>

[Image Source](https://en.wikipedia.org/wiki/Confederation_Bridge)

For most of engineering history, assessing whether a bridge remained safe once errected required periodic visual inspections to catch problems like fatigue, corrosion, or ice damage before they became dangerous.

Structural engineers now increasingly use sensor data and data science tools to address this question.

With sensors (like strain gauges, thermal sensors, tiltmeters, vibration sensors, etc.), structural engineers can continuously monitor the health of bridges and detect early warning signs of structural deterioration or failure by analyzing this constant stream of measurements. 

For example, the Confederation Bridge (the 12.9km link between Prince Edward Island and New Brunswick designed for a 100-year service life) has hundreds of sensors embedded in it. 

This approach, known as structural health monitoring, is being used worldwide in bridges, dams, and buildings.

### 💊 Every Cure: AI and the Search for Hidden Drug Uses

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/every_cure.png" alt="every cure" width="800"/>
</div>

[Image Source](https://everycure.org/)

Of the roughly 22,000 diseases recognized by medicine, only about 3,000 have an FDA-approved treatment.

BUT, many of the drugs already approved for one condition could plausibly treat others, a phenomenon known as _drug repurposing_. 

Until now, this has mostly happened by chance. This is what happened for a physician named David Fajgenbaum who found an "off-label" use of sirolimus (a drug originally developed as an immunosuppressant for use after kidney transplants) to treat his own Castleman disease.

To systematize this chance finding, he co-founded the nonprofit Every Cure to use AI to evaluate the efficacy of every known drug to every known disease. There are 66 million such drug-disease matches.

The organization has already advanced dozens of repurposed drug candidates and aims to bring 15 to 25 newly repurposed treatments to patients by 2030, especially targeting the thousands of rare diseases that pharmaceutical companies have little financial incentive to research on their own.

### 📈 Finance and Stock Prediction: The GameStop Short Squeeze

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/gme_price.png" alt="gme price" width="800"/>
</div>

Modern finance runs heavily on data science: quantitative hedge funds build statistical models on historical prices, trading volumes, and company fundamentals in an effort to predict which stocks will rise or fall. 

Recently, it has become clear that public sentiment on Reddit is also a powerful predictor of whether a stock price will rise or fall.

In January 2021, retail investors coordinating on the subreddit r/WallStreetBets were responsible for the _short squeeze_ of the GameStop stock: share price rocketed from around \\$17 in early January to an intraday high near \\$483 by January 28.

Since then, financial analysts and data scientists have been much more diligent to incorporate information found on Reddit threads in their predictive models.
