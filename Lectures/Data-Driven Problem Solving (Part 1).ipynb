## The Scientific Method (and why it's not enough)

The **Scientific Method** has been the key to humanity understanding ourselves and the world around us. It's rooted in *skepticism* and *empiricisim*; you shouldn't trust something unless there is evidence to validate it. It looks something like this:

<div style="text-align: center;">
<img src="https://thumb.wikimedia.org/wikipedia/commons/thumb/8/82/The_Scientific_Method.svg/500px-The_Scientific_Method.svg.png" alt="Scientific Method" width="500"/>
</div>

[Image Source](https://en.wikipedia.org/wiki/Scientific_method)

This framework is useful at a high level, and it explains how scientific theories have been developed, tested, and validated over time. 

But it *does not* provide a usable set of instructions for the person who has to actually run a study to answer a particular question.

Notice what the scientific method does not tell you:

- *Who or what* am I measuring, exactly?
- How do I know the thing I can measure is the thing I care about?
- How many observations do I need?
- What do I do about all the other stuff that varies and that I'm not interested in?
- How do I know my measuring instrument isn't lying to me?
- What claim am I entitled to make at the end, and how confident should I be?

**PPDAC: *Problem, Plan, Data, Analysis, Conclusion* is a framework that *does* answer those questions.**

## The Statistical Method (PPDAC)

### The Purpose

PPDAC was originally proposed by UWaterloo professors Jock MacKay and Wayne Oldford in [this paper](jstor.org/stable/2676665), and they originally referred to it as _The Statistical Method_.

It turns out, _The Statistical Method_ is broadly useful in the field of data science as well. 

In particular, **PPDAC provides a formal five-stage structure for data-driven problem solving** and it has two main uses:

* it's a *template* for actively running your own investigation, and 
* it's a *checklist* for critically reviewing someone else's. 

We'll use it both ways in this course, and the second use is one that will serve you for the rest of your life as you become critical consumers data and information.

### The Structure

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/58fc1f744c0caf95ae25e32836bb8625c59a40b7/images/PPDAC.svg" alt="PPDAC" width="800"/>
</div>


* **Problem:** a clear statement of the study's objectives, usually involving one or more questions.
* **Plan:** the procedures used to carry out the study including how we will collect the data.
* **Data:** the physical collection of the data, as described in the Plan.
* **Analysis:** the analysis of the data collected in light of the Problem and the Plan.
* **Conclusion:** The conclusions that are drawn from the Analysis about the Problem and their limitations.

Two rules govern these arrows:

* **Looking backward:** each stage is *legitimized* by the stages before it. A Plan that doesn't address the Problem is worthless — and if that happens, one of the two has to change. An Analysis that ignores how the Data were collected is not an analysis; it's some meaningless calculations.

* **Looking forward:** choices made early make life easier later. A well-designed Plan can make the Analysis almost trivial. A badly designed Plan can make the Analysis impossible, no matter how clever you are.

### The Relationship with The Scientific Method

> *"Isn't PPDAC just the scientific method with different words?"*

Fair question. The answer is no. PPDAC is *finer-grained* and *operational*, and it's scoped to a single investigation and a specific problem of interest.

Here's a summary of some important differences:

| | Scientific Method | Statistical Method (PPDAC) |
|---|---|---|
| **Scope** | A field of knowledge, over decades | One investigation, over days weeks or months |
| **Unit of progress** | A theory | A study |
| **Driven by** | Explanation, mechanism | Measurement, variation, uncertainty |
| **Answers** | "Why is the world like this?" | "What can *these* data support, and how sure can I be?" |
| **Who does it** | A community | A team, or a person |

**Key Takeaway:** The Scientific Method may require many individual investigations to test and validate a theory. Each of those individual investigations requires The Statistical Method (PPDAC) to carry out effectively.

## PROBLEM — What, exactly, are we trying to learn?


> *"Understanding what is to be learned from an investigation is so important that it is surprising that it is rarely, if ever, treated in any introduction to statistics."*
> — MacKay & Oldford (2000)

This quote highlights shortcomings of traditional education in statistics and data science. Often, the problem is solved to begin with: you're handed a clean data set and told which analysis to perform. Real work never looks like that. Real work starts with a vague sentence from a person who is not a data scientist.

**The purpose of the Problem stage is to turn that vague sentence into a precise statement of what is to be learned.**

### A taxonomy of data science problem types

Every data science project begins with a question. There are many kinds of questions data scientists may answer, and in many different fields. But they can often be classified as one of the six following types: _descriptive_, _exploratory_, _predictive_, _inferential_, _causal_, or _mechanistic_. We summarize these question types in the table below, which is reproduced from a simular table given in our textbook [Data Science: A First Introduction with Python](https://python.datasciencebook.ca/intro.html#asking-a-question).

| Type | What it asks | Canonical example | 
|---|---|---|
| **Descriptive** | Summarized characteristics of a data set, without interpretation (e.g., reporting a fact). | How many people live in each province and territory in Canada? | 
| **Exploratory** | Are there patterns, trends, or relationships *within this data set*? Often generates hypotheses for future study. | Does political party voting change with indicators of wealth among these 2,000 Canadians? | 
| **Predictive** | What value or label should we expect for a *new* individual? Focuses on what *predicts* an outcome, not what *causes* it. | Which party will this person vote for in the next election? | 
| **Inferential** | Is there a pattern *and* does it generalize to the wider population? The generalization is quantified. | Does political party voting change with wealth for *all* people living in Canada? | 
| **Causal** | Would *changing* one factor change another, on average, in the wider population? | Does wealth *lead to* voting for a certain party? | 
| **Mechanistic** | *How* does it happen? What is the underlying mechanism? | *How* does wealth lead to voting for a certain party? | 

**Two important distinctions to make:**

* *Exploratory vs. inferential:* Same pattern, different claim. 
    * Example: exploratory says "in *this* data set, wealthier respondents voted differently." Inferential says "and here is how confident I am that this holds for the *whole country*, $\pm$ this much." The difference is entirely about whether you quantify the generalization.
* *Predictive vs. causal:* Predictive asks *what travels with the outcome*. Causal asks *what moves the outcome*. 
    * Example: Hospital data will tell you that patients on ventilators are much more likely to die. That may be an accurate *prediction*, but a problematic *causal* claim.
    
**Knowing the type of question you're answering is the most important part of the entire investigation**

The question type determines:
* what data you need,
* what analysis is relevant, and 
* what conclusions are appropriate.

### The vocabulary of a well-posed problem

| Term | Definition | 
|---|---|
| **Unit** | The individual thing you measure. | 
| **Target population** | The whole collection of units you want to draw conclusions about. | 
| **Response variable** | The characteristic of each unit you primarily care about. | 
| **Explanatory variables** | Other characteristics of each unit that might explain why the response varies. | 
| **Population attribute** | A characteristic of the *whole* target population that you'd like to know about. |

### Example 1 — 🎬 Netflix

From the Introduction lecture, recall Netflix's recommendation system and its culture of A/B testing.

Imagine a manager says: *"I want to understand our recommendation system better."* That's not a question, it's a statement. But clearly, the manager has a question in mind. There is something they want to know. 

Here are six possible questions, each a different type:

| # | Question | Type | Why |
|---|---|---|---|
| 1 | What fraction of hours watched on Netflix last quarter started from a row on the home screen rather than from search? | **Descriptive** | A fact about a data set. No model, no generalization, no claim about mechanism. |
| 2 | Do users who watch a lot of documentaries also tend to watch a lot of true crime? | **Exploratory** | Looking for a relationship *within* the data at hand. Could generate a hypothesis. |
| 3 | Given everything we know about this user, which five titles should we show them right now? | **Predictive** | Predicting viewing preferences for an individual unit. We do not care *why* it works. |
| 4 | Among all Netflix subscribers worldwide, is the completion rate for limited series higher than for multi-season shows, and by how much? | **Inferential** | A pattern *plus* a quantified generalization to the full subscriber population. |
| 5 | Does showing thumbnail art featuring a human face cause a higher click-through rate? | **Causal** | The word "cause" indicates we're asking what happens if we *intervene*. |
| 6 | *Why* does human-face artwork increase clicks — is it facial recognition drawing the eye, an inference about genre, or something else? | **Mechanistic** | Asking for the machinery, not just the effect. |

**Let's dig into Question 5**

The Problem statement can be made more precise with the PPDAC vocabulary:

- **Unit:** one Netflix user who sees a title's thumbmail art.
- **Target population:** all Netflix users who see a title's thumbmail art.
- **Response variable:** upon seeing the title's thumbmail art, did the user click to watch the title (yes/no)? Possibly also: did the user watch the title for more than two minutes? 
- **Explanatory variables:** thumbnail art variant (face vs. no-face), title, genre, users's viewing history, device type, row position on screen, time of day, country.
- **Attribute:** the difference in click-through rate between the face and no-face variants, across all Netflix users who see thumbnail art.

Notice how much work that was, and notice that we haven't collected or analyzed any data yet. Much of a data-driven investigation happens before any data is collected. This is normal

### Example 2 — 🦠 COVID-19 in Canada

From the Introduction lecture, recall the CanCOGeN genomic surveillance, variant modelling, and wastewater monitoring.

Imagine the Premier of a Province says: *"Tell me what's going on with COVID."* Again, that is not a question, but the Premier has _some_ question in mind. There is something they want to know.

Here are six possible questions, each a different type:


| # | Question | Type |
|---|---|---|
| 1 | How many COVID-19 cases were reported in each province last week? | **Descriptive** |
| 2 | Does the wastewater viral signal in a city tend to rise before that city's reported case counts do? | **Exploratory** |
| 3 | Given this week's wastewater concentration, how many hospitalizations should we expect 10 days from now? | **Predictive** |
| 4 | What proportion of *all* Canadians had been infected by the end of 2021 (not just those who got a positive test)? | **Inferential** |
| 5 | Did the province-wide mask mandate reduce transmission? | **Causal** |
| 6 | *How* does the Omicron spike protein's mutation profile allow it to escape prior immunity? | **Mechanistic** |

**Important lessons this example highlights**

* **Target vs Study Population:** Question 4 asks about *all Canadians infected*. But during the Omicron wave, provincial testing capacity collapsed and PCR tests were restricted to healthcare workers and high-risk groups. So the population you can actually observe through case counts is "people who got a test", which is not representative of "people who got infected". The latter is the target population, and the former, the units we actually have access to, is called the **study population**. When there is a gap between those two, we encounter **study error**, which may render analyses and resulting conclusions meaningless.

* **Descriptive is not the same as easy:** Question 1 asks "How many cases last week?" This sounds trivial, but it's not: is it cases reported by date of test or date of symptom onset? Does a reinfection count? Do rapid tests count, and are they even reported? Is a "case" a person or an episode? Considerations like these need to be carefully addressed even in what seemed like the simplest question on the list.

## PLAN — How will we get data that can actually answer this?


> *"There is little value in a Plan that does not address the Problem; in such a case, one of the two stages must be modified."*
> — MacKay & Oldford (2000)

The Plan stage is where the Problem meets reality. Its purpose is to decide **how the data will be collected and analyzed**. It is the stage that most distinguishes a data scientist from someone who merely knows how to run models.

### Target Population, Study Population, Sample, and a tale of three errors

When Planning an investigation, it's imperitive to consider how the data we Analyze is related to the target population specified by the Problem. The schematic below illustrates this relationship.

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/a7c491d3909fe94e8a81af5b383723a7dd612063/images/target-study-sample.svg" alt="target-study-sample" width="500"/>
</div>

**Recall:**

* The **target population** is the collection of units we would like to draw conclusions about.
* The **study population** is the collection of units we're actually able to study. Sometimes this is the same as the target population.
* And **study error** exists whether there is a meaningful difference between these two. No amount of data is going to fix this.
    * *Example:* Most of what psychologists have learned about human behaviour has been obtained by studying WEIRD people (Western, Educated, Industrialized, Rich, Democratic). For a more thorough treatment of this example of study error, see [Henrich (2020)](https://heb.fas.harvard.edu/publication/weirdest-people-world-how-west-became-psychologically-peculiar-and-particularly):

<div style="text-align: center;">
<img src="https://heb.fas.harvard.edu/sites/g/files/omnuum1771/files/styles/hwp_4_5__480x600/public/2026-03/WEIRDest%20people%20henrich.jpg" alt="WEIRD" width="300"/>
</div>

[Image Source](https://heb.fas.harvard.edu/publication/weirdest-people-world-how-west-became-psychologically-peculiar-and-particularly)

> *"Behavioral scientists routinely publish broad claims about human psychology and behavior in the world's top journals based on samples drawn entirely from Western, Educated, Industrialized, Rich, and Democratic (WEIRD) societies. Researchers - often implicitly - assume that either there is little variation across human populations, or that these "standard subjects" are as representative of the species as any other population. Are these assumptions justified? Here, our review of the comparative database from across the behavioral sciences suggests both that there is substantial variability in experimental results across populations and that WEIRD subjects are particularly unusual compared with the rest of the species - frequent outliers. The domains reviewed include visual perception, fairness, cooperation, spatial reasoning, categorization and inferential induction, moral reasoning, reasoning styles, self-concepts and related motivations, and the heritability of IQ. The findings suggest that members of WEIRD societies, including young children, are among the least representative populations one could find for generalizing about humans. Many of these findings involve domains that are associated with fundamental aspects of psychology, motivation, and behavior - hence, there are no obvious a priori grounds for claiming that a particular behavioral phenomenon is universal based on sampling from a single subpopulation. Overall, these empirical patterns suggests that we need to be less cavalier in addressing questions of human nature on the basis of data drawn from this particularly thin, and rather unusual, slice of humanity. We close by proposing ways to structurally re-organize the behavioral sciences to best tackle these challenges."* - [Henrich et al., (2010)](https://pubmed.ncbi.nlm.nih.gov/20550733/)

**But we don't usually have access to full study population**

* Instead, the data we collect is on a subset of study population units. We call this subset the **sample**. 
* If there is a meaningful difference between the sample and the study population, we encounter **sample error**. 
* Sample error is reduced by ensuring the sample is *representative* of the study population. 

**And even if our sample is representative of the study population, and that's representative of the target population, we may still have problems**

* **Measurement error** is the difference between what your measuring process reports and the true value. 
    * _Example:_ Diagnostic tests are often associated with errors - false positives and false negatives. If a Province's Premier wants to know how many COVID-19 cases there were last week, and the measurements we have are based on PCR test results, we have to acknowledge that the *observed* case count may differ from the *true* case count because of measurement error. Too many false positives will bias our observed case count upward, and too many false negatives will bias our observed case count downward.

### Data Collection Strategies

The data you analyze may come from a number of different sources. Here are some common methods of collecting data.

| Strategy | What it is |  Example |
|---|---|---|
| **Found data** | Data that already exists, collected for some other purpose entirely. | DECIMA Florentine tax censuses |
| **Digitial trace data** | Digital records left behind when people interact with online systems. | Reddit posts used to predict GameStop |
| **Observational study** | You deliberately go and measure units, but you do not manipulate anything. | Kepler staring at 200,000 stars |
| **Sample survey** | You define a target population of units, draw a sample from it, and ask them questions. | Election polling |
| **Sensor data** | Continuous automated measurement of a process. | Confederation Bridge strain gauges |
| **Controlled experiment** | You investigate a causal relationship by controlling explanatory variates to observe the impact on the response. | Netflix A/B tests |
| **Simulation data** | You generate data from a computer model of a system. | Climate models |
| **Meta-analysis** | Combining many prior studies. | Combining clinical trial results |

A few remarks:

* **"Found data" is the default in modern data science, and it's the most dangerous.** Almost everything in the Introduction lecture that feels modern — Reddit sentiment, streaming logs, satellite imagery — is found data. It is cheap, enormous, and *collected by someone else for some other purpose*. You did not choose the sampling protocol, so tread carefully.

* **Simulation deserves a special warning.** A climate model or an epidemic model produces data with perfect control and infinite sample size. But the target population is the *real world* and the study population is *the model*. The study error is the model's wrongness. This is why the NASA result cited in our Introduction lecture (that past climate model projections have matched observed warming well) is such a big deal: it is an empirical check on study error.

* **Controlled experiments and causal inference.** Two variables are said to be causally related if changing one causes the other to change in some way. Only with a controlled experiment can we investigate causal relationships because it's the only data collection strategy in which the investigators exert control in the data collection process. If you collected observational data and someone asks a causal question, the correct answer is *"these data cannot answer that,"* not *"let me fit a different model."*


### Matching the Plan to the Problem

Your data collection strategy must align with your Problem type; some data collection strategies are not appropriate for answering certain question types.

| Question type | What you need from the Plan | Appropriate data collection strategy | 
|---|---|---|
| **Descriptive** | Clear definitions and measurements on one or more variables. | Almost anything, *if* you're honest that you're describing only the data at hand. |
| **Exploratory** | Enough variables, enough units, and rich enough structure to reveal patterns. | Found data, observational study, sensor data, digital trace. | 
| **Predictive** | Data whose relationships resemble the setting where predictions will be made. | Found data, observational study, sensor data, digital trace. | 
| **Inferential** | A random sample carefully taken from a broader (target or study) population. | Sample survey, observational study. | 
| **Causal** | Manipulation of explanatory variables and observation of resulting impact on response variable. | Controlled experiment | 
| **Mechanistic** | Measurement of intermediate steps; usually a theoretical model; usually many nested studies. | Experiment + theory + instrumentation, iterated. | 


**Key takeaway:** The question type determines which data collection strategies are appropriate.
