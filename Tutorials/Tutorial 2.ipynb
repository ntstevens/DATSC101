## 📚 Libraries and Functions

We've just about exhausted the mathematical operations "base" Python is capable of. When we need to perform a more advanced calculation, we're going to need to import a **library**.

What is a library?
* 🧙 Think of yourself as a wizard capable of many advanced spells, but you need your spellbook in order to do them. 
* 🧑‍🍳 Think of yourself as a chef capable of many advanced recipes, but you need your cookbook in order to do them.
* 🧑‍🔬 Data scientists are similar. We can apply many advanced *functions*, but we need access to the library containing those functions in order to use them.

What is a function?

* A function is a reusable, self-contained block of code designed to perform a specific task.

Today we'll work with a library called `numpy` (pronounced *numb-pie*), which is commonly used for scientific computing in Python. We'll use is to perform some mathametical calculations available to us via functions.

## 💽 Loading Data

Data scientists analyze data, and they do so with Python. So, the first step in any data analysis is to load the data into your working environment so that Python has access to it. We do this by _loading_ or _reading in_ the data.

---

### Tabular Data

Data comes in a variety of shapes, sizes, and formats, but we will begin with the most basic data structure: **tables**. Tabular data is what you think of if you imagine information stored in spreadsheets. It's rectangular in shape, with **rows** and **columns**. The image below provides an example of tabular data containing information about movies. 

<div style="text-align: center;">
<img src="https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/images/movie_info.png" alt="movie info" width="800"/>
</div>


The top _row_ of this dataset is a **header** row which provides the _column_ names. We can see that the columns represent movie attributes.

Every other row represents a different movie, and the columns contain specific characteristics for the respective movie. In particular, we have information on:

* `names`: the movie name
* `date_x`: the release of the movie
* `score`: the IMBD score for the movie (multiplied by 10)
* `genre`: the genre(s) of the movie
* `overview`: a summary of the movie
* `crew`: notable cast in the movie
* `orig_title`: the original tile of the movie (often but not always the same as `names`)

**In general, tabular data is structured as follows:**

* Rows represent the objects we study. We call these **observations**.
* Columns represent characteristics of those objects. We call these **variables**.
* A **cell** contains the value of a particular variable for a particular observation.

---
### Introduction to `pandas`

So, if we have a dataset like this, stored in a spreadsheet, how do we load it? With pandas! 

No, not these _pandas_: 🐼

We use `pandas`. It's another library like `numpy` which contains helpful functions. Here we will illustrate a particularly useful function for reading in tabular data:

* `read_csv` which reads in tabular data stored in a `.csv` file

#### Importing and using `pandas`

Let's import `pandas` and start to get familiar with these functions. Note that we'll important `pandas` with the commonly used alias `pd`.

````python
import pandas as pd
````

The dataset pictured above is stored in a `.csv` file called `imdb_movie_info.csv`. With that file in hand, we can read it in with the function `read_csv`. But first, recall the syntax needed to use a function contained in a library:

```Python
library.function("argument")
```

So, we'll need to do something like: 

```Python
pd.read_csv(...)
```

But obviously we need to replace `...` with something. _But what?_

#### Accessing Documentation

Now is a good time to illustrate the use of **documenation**; a function's documentation tells you how to use the function. It explains all of the relevant inputs and outputs. We can access it with the `help` function or by typing the question mark `?` ahead of a function name. Try it below:

````python
? pd.read_csv
````

As we can see, `pd.read_csv` takes many inputs. While we'll explore many of them later in the course, today we'll focus on the main one. The first one:

```
filepath_or_buffer : str, path object or file-like object
    Any valid string path is acceptable. The string could be a URL. Valid
    URL schemes include http, ftp, s3, gs, and file. For file URLs, a host is
    expected. A local file could be: file://localhost/path/to/table.csv.
```

So, if we pass the filepath as a string, `pd.read_csv` will 

1. know what we're looking for,
2. know where to find it, and 
3. read it in. 

#### Loading data from the same folder as your notebook

Let's try it out:

````python
pd.read_csv("imdb_movie_info.csv")
````

Notice that we simply passed the name of the file. This only worked because the file was saved in the same folder as this Jupyter notebook. If it was stored somewhere else, we would have needed to pass a more specific path name. 

#### Loading data from a different folder than your notebook
Suppose instead that the folder containing the notebook also contained a folder called `data`, which itself contained the data file `imdb_movie_info.csv`. In that case we should use the following command instead:

````python
pd.read_csv("data/imdb_movie_info.csv")
````

In general: we find the `filepath` for the data file, and then run:

```Python
pd.read_csv("filepath")
```

#### Loading data from a webpage

Sometimes the data we want to access is stored on the web. Thankfullty, can also use `read_csv` to easily load a csv file directly from a website. We simply need to pass the webpage's URL as the `filepath`.

For instance, the `imdb_movie_info.csv` file is also stored on our course's GitHub page.

* The ordinary csv file can be found here: https://github.com/ntstevens/DATSC101/blob/main/data/imdb_movie_info.csv
* And the "raw" version can be found here: https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/data/imdb_movie_info.csv

It's the URL for the raw version that we want to use:

````python
pd.read_csv("https://raw.githubusercontent.com/ntstevens/DATSC101/refs/heads/main/data/imdb_movie_info.csv")
````

#### What if my data is not in a `csv` file?

Tabular data is not necessarily store in `csv` files. It may also be stored in:

* `.xlsx` files
* `.json` files
* `.txt` files
* ...

But no problem, `pandas` provides functionality for reading in data from many sources. We're only getting started. We'll gain experience loading data from other sources later in the course.
