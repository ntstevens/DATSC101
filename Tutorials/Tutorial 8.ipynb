## The Setup

In Tutorial 5 we explored to different *sampling mechanisms*:
* **Judgement Sampling**: the investigator selects units they judge to be representative of the population.
* **Simple Random Sampling**: every unit is selected at random, and every unit in the population has the same chance of being selected.

We found the sampling mechanism had an impact on the quality of samples produced; simple random sampling was associated with smaller and less variable sample error.

Today we'll revisit the population of 100 plastic blocks of various shapes and sizes, for which the *average weight* of all 100 blocks is the attribute of interest. In particular, we will investigate the effect of sampling mechanism on the performance of confidence intervals and hypothesis tests for the average block weight.
