# Introduction to Cells

As you will see, a notebook is composed of a collection of "cells". These cells can be interpreted in one of three ways:

* **Code:** As the name suggests, a code cell is one with which you write and execute code. In DATSC101, Python is the programming language we'll be using, and so our notebooks will be setup such that the code cells expect Python code. 
* **Markdown:** Markdown is a markup language that uses simple keyboard symbols to create formatted text. Anytime you want to type text that is not code, it will be in a Markdown cell. 
* **Raw:** A cell type whose content is completely ignored by the notebook kernel and never executed. We won't make use of these in DATSC101.

This text you're reading right now is stored in a cell. A Markdown cell to be specific. See for yourself. Double-click the text and you will see the nicely formatted text revert back to it's representation in the plain-text editor. The plain-text version is the source, and when the cell is "Run", the text is rendered nicely with the desired formatting.

## Running a Cell

How do you "Run" a cell? A few ways:

* You can click the play button (&#x25B6;) at the top of the notebook.
* You can click the "Run" menu from the top of the JupyterHub application, and from there you can choose to run one or more cells.
* You can use a keyboard shortcut
    * Mac:
        * Run cell in place: Command + Return
        * Run cell and insert below: Option + Return
        * Run cell and select below: Shift + Return
    * Windows
        * Run cell in place: Ctrl + Enter
        * Run cell and insert below: Alt + Enter
        * Run cell and select below: Shift + Enter

As you may have noticed, newly created cells are by default code cells, but you can change this. To change a given cell's type, put your cursor in the cell and select the desired type from the dropdown menu at the top of the notebook.

# Introduction to Markdown

At this point in your studies, you are undoubtedly familiar with word processors like Microsoft Word or Google Docs. In those environments, text formatting involves highlighting text and clicking a button. Here, you will gain experience with Markdown, which only requires keyboard symbols to format text.

---
## 📝 Text Basics

* **Bold text:** Wrap words in two asterisks `**like this**` to make them **like this**.
* **Italic text:** Wrap words in single asterisks `*like this*` to make them *like this*.
* **Strikethrough:** Wrap words in two tildes `~~like this~~` to make them ~~like this~~.
* **Underlined text:** Unfortunately, there is no built-in syntax to underline text in Markdown.
* **Combined emphasis:** Wrap words in three asterisks `***bold and italic***` for ***bold and italic***.

## 🏛️ Headings and Structure

Instead of selecting "Heading 1" or "Heading 2" from a dropdown menu, use the hash # symbol at the start of a line. Always put a space after the hash symbols.

For example, this:

```
# Title (Heading 1)
## Major Section (Heading 2)
### Subsection (Heading 3)
#### Sub-subsection (Heading 4)

```
renders as this:

# Title (Heading 1)
## Major Section (Heading 2)
### Subsection (Heading 3)
#### Sub-subsection (Heading 4)

## 📋 Lists

Markdown formats lists automatically when you start a line with standard listing characters.

### Bulleted Lists

Use a dash `-` or an asterisk `*` followed by a space. Indent two spaces for sub-bullets.

For example, this:

```
* First main point
* Second main point
  * Sub-point under the second point
```
randers as this:

* First main point
* Second main point
  * Sub-point under the second point
  

### Numbered Lists

Type the number followed by a period and a space. Markdown automatically fixes the numbering sequence when rendered, even if you type 1. for every item.

For example, this:

```
1. First step
2. Second step
1. Third step (Markdown will correctly display this as 3.)
```
renders as this:
1. First step
2. Second step
1. Third step (Markdown will correctly display this as 3.)

## 💻 Faux Code

Word processors often struggle to format computer code neatly. Markdown makes it simple.

### Inline Code

To highlight a variable name, file path, or short command inside a sentence, wrap it in single backticks `` ` `` (found next to the `1` key).
* Example: Use the `print()` function.

### Code Blocks
To display a multi-line block of programming code, wrap the entire block in triple backticks `` ``` ``. 

```
def hello_world():
    print("Hello, student!")
```

You can type the language name right after the first set of backticks to enable color coding.

```python
def hello_world():
    print("Hello, student!")
```

To be clear, we aren't writing code that can be executed here; we're simply formatting text to look like it's code. If we'd like the code to run, we should change the cell type.

## 📐 Math Formulas

For math equations, Jupyter uses LaTeX (pronounced _lay-tech_) syntax. Wrap your math in dollar signs `$` instead of using a complex equation editor.

### Inline Math

Wrap equations in single dollar signs `$` to keep them in the sentence line.
* `$E = mc^2$` renders as $E = mc^2$.

### Centered Equations

Wrap equations in double dollar signs `$$` to center them on their own line.
* `$$\frac{1}{2}$$` creates a centered fraction: $$\frac{1}{2}$$

For a fairly comprehensive overview of LaTeX syntax, see [this cheatsheet](https://webhome.auburn.edu/~tamtiny/Symbols.pdf).
