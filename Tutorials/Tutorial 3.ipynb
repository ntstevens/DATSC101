Today we'll begin working with it. As we saw, the data set contains information from IMDb (Internet Movie Database) about many movies. An overview of the variables recorded for each movie is provided below.

* `names`: the movie name
* `date_x`: the release of the movie
* `score`: the IMBD score for the movie (multiplied by 10)
* `genre`: the genre(s) of the movie
* `overview`: a summary of the movie
* `crew`: notable cast in the movie
* `orig_title`: the original tile of the movie (often but not always the same as `names`)
* `status`: the production status of the movie (either  `In Production`, `Post Production` or `Released`)
* `orig_lang`: the language the movie was originally released with
* `budget_x`: the budget (in USD) of the movie
* `revenue`: the revenue (in USD) of the movie
* `country`: the country the movie was originally released in

#### 3. Confirm that the variable names listed above are correct.

#### 4. Rename `date_x` and `budget_x` to exclude `_x`.

```python
movies = movies.rename(columns = {"date_x":"date",
                                  "budget_x":"budget"})
movies.columns
```
