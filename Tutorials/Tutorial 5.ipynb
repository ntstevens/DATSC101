## The Setup

*Sampling* is a powerful and widely used tool to learn about a *population attribute* without examining every unit in the population. Examples include public opinion polls such as the Gallup poll, the Labour Force Survey to establish unemployment rates, market research surveys to assess a new product and so on.

One important aspect in sampling is the method used to select the sample (also known as the *sampling mechanism*). In this tutorial, you will compare two different sampling mechanisms.

The population is a collection of 100 plastic blocks of various shapes and sizes. The attribute of interest is the *average weight* of all 100 blocks. 

We will investigate our ability to estimate this population average with data collected from samples of size $n=10$.
